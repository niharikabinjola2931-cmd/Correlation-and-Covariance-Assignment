### Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.

Ans1. Covariance measures how two numerical variables change together. It indicates the direction of their linear relationship: whether they increase together (positive) or move in opposite directions (negative).

1. On the basis of Scale
 
 Covariance is unstandardized and depends on the units of the variables. If you measure weight in kilograms versus grams, the covariance value changes drastically, making it impossible to compare relationships across different datasets.
 
Correlation is dimensionless and standardized. It is calculated by dividing the covariance by the product of the variables' standard deviations, which removes the units entirely

2. On the basis of Interpretation
 
 Covariance Interpretation provides the direction and the raw magnitude of joint variability. While a positive value means they move together and a negative value means they move in opposite directions, the exact numerical value holds no absolute limit—it can range from (-infinity) to (+infinity), meaning you cannot easily determine how strong the relationship is just by looking at the number.

Correlation Interpretation measures both the direction and the strength of the linear relationship. The resulting coefficient is strictly bounded between -1 and 1. A value closer to \(\pm 1\) indicates a strong linear relationship, while a value near \(0\) indicates a weak or non-existent linear relationship

### Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables?

Ans2. A positive, negative, or zero covariance indicates the directional relationship between two variables.

Three Types of Covariance

Positive Covariance: Variables move in the same direction. When variable A increases, variable B tends to increase. When variable A decreases, variable B tends to decrease.

Negative Covariance: Variables move in opposite directions. When variable A increases, variable B tends to decrease.

Zero Covariance: Variables have no linear relationship. A change in variable A does not predict a predictable upward or downward movement in variable B.

### Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation
preferred in many cases?

Ans3. Limitations of Covariance

 Covariance changes when measurement units change. Measuring height in meters versus centimeters yields completely different covariance values.
 
Values range from negative infinity to positive infinity.

You cannot determine the strength of a relationship from the raw number alone. A covariance of 100 is not inherently stronger than a covariance of 1

It only captures linear relationships. Variables with a perfect curvilinear relationship can still result in a covariance of zero.

Extreme data points disproportionately distort the results.

Why Correlation is Preferred

Standardized Scale: Correlation forces the relationship metric onto a strict scale from -1 to 1.

Unitless Metric: The calculation strips away all units of measurement. You can directly compare the relationship of height vs. weight to the relationship of income vs. education.

### Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation
coefficient. When would you prefer to use Spearman’s correlation?

Ans4. Difference between Pearson's correlation coefficient and Spearman's rank correlation coefficient

Data Type: Pearson measures linear relationships between continuous, interval, or ratio variables. Spearman measures monotonic relationships using ordinal (ranked) or transformed continuous data.

Relationship Type: Pearson evaluates if variables change at a constant rate (a straight line). Spearman evaluates if variables change together, even if the rate is not constant (curves up or down monotonically).

Calculation Basis: Pearson uses the raw, actual values of the data points. Spearman strips away the raw values and uses only the relative rank order of the data points.

Outlier Sensitivity: Pearson is highly sensitive to outliers, which can easily distort the coefficient. Spearman is robust against outliers because extreme values are capped at the highest or lowest ranks

.When to Prefer Spearman’s Correlation

Non-Linear but Monotonic Data: Use it when your data creates a curve rather than a straight line, but both variables still consistently increase or decrease together.

Ordinal Data Types: Use it when analyzing ranked data, such as satisfaction surveys (e.g., Likert scales from 1 to 5) or competition placements (1st, 2nd, 3rd).

Presence of Outliers: Use it when your dataset contains extreme anomalies that you cannot remove, as ranking prevents these points from heavily skewing the results.

Violated Normality: Use it when your continuous data is heavily skewed and fails normality assumptions required for parametric tests.
